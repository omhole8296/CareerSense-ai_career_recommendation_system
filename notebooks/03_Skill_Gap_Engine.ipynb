```python
import pandas as pd
import numpy as np

career_skills = pd.read_csv("career_skills.csv")

career_skills.head()
```

```python
career_skills.shape
```

```python
student_skills = {
    "Python": 4,
    "SQL": 3,
    "Excel": 4,
    "Statistics": 3,
    "Data Analysis": 4,
    "Machine Learning": 1,
    "Deep Learning": 0,
    "NLP": 0,
    "Power BI": 2,
    "Tableau": 2,
    "Java": 0,
    "JavaScript": 2,
    "HTML_CSS": 2,
    "C++": 1,
    "Git": 3,
    "Linux": 1,
    "Cloud": 0,
    "Cybersecurity": 0,
    "Communication": 4,
    "Problem Solving": 4
}
```

```python
target_career = "Data Scientist"
career_row = career_skills[
    career_skills["Career"] == target_career
].iloc[0]

career_row
```

```python
def calculate_skill_gap(student_skills, career_row):

    skill_gap = {}

    for skill in student_skills:
        required = career_row[skill]
        current = student_skills[skill]

        gap = required - current

        skill_gap[skill] = {
            "Current": current,
            "Required": required,
            "Gap": gap
        }

    return pd.DataFrame(skill_gap).T
```

```python
gap_df = calculate_skill_gap(
    student_skills,
    career_row
)

gap_df
```

```python
missing_skills = gap_df[
    gap_df["Gap"] > 0
].sort_values(
    by="Gap",
    ascending=False
)

missing_skills
```

```python
def categorize_gap(gap):

    if gap >= 3:
        return "High Priority"

    elif gap == 2:
        return "Medium Priority"

    elif gap == 1:
        return "Low Priority"

    else:
        return "No Gap"
```

```python
gap_df["Priority"] = gap_df["Gap"].apply(
    categorize_gap
)

gap_df
```

```python
strong_skills = gap_df[
    gap_df["Gap"] <= 0
].sort_values(
    by="Current",
    ascending=False
)

strong_skills
```

```python
required_values = gap_df["Required"].replace(0, np.nan)
current_values = gap_df["Current"]

skill_percentages = (
    current_values / required_values
) * 100

skill_percentages = skill_percentages.clip(
    upper=100
)

readiness_score = skill_percentages.mean()

print(
    "Career Readiness Score:",
    round(readiness_score, 2)
)
```

```python
print("================================")
print("       CAREERSENSE REPORT")
print("================================")

print("\nTarget Career:")
print(target_career)

print("\nCareer Readiness Score:")
print(round(readiness_score, 2), "%")

print("\nHigh Priority Skills:")
print(
    list(
        gap_df[
            gap_df["Priority"] == "High Priority"
        ].index
    )
)

print("\nMedium Priority Skills:")
print(
    list(
        gap_df[
            gap_df["Priority"] == "Medium Priority"
        ].index
    )
)

print("\nStrong Skills:")
print(
    list(
        gap_df[
            gap_df["Gap"] <= 0
        ].index
    )
)
```
