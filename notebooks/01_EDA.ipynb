```python
plt.figure(figsize=(10, 5))

df["Target_Career"].value_counts().plot(kind="bar")

plt.title("Student Distribution by Career")
plt.xlabel("Career")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
skill_columns = df.columns[1:-1]

skill_means = df[skill_columns].mean().sort_values(ascending=False)

skill_means
```

```python
plt.figure(figsize=(12, 6))

skill_means.plot(kind="bar")

plt.title("Average Skill Level Across Students")
plt.xlabel("Skill")
plt.ylabel("Average Proficiency")
plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

```python
career_skill_means = df.groupby("Target_Career")[skill_columns].mean()

career_skill_means
```

```python
plt.figure(figsize=(14, 8))

sns.heatmap(
    career_skill_means,
    annot=True,
    fmt=".1f"
)

plt.title("Average Skill Profile by Career")
plt.xlabel("Skills")
plt.ylabel("Career")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    df[skill_columns].corr(),
    cmap="coolwarm"
)

plt.title("Skill Correlation Matrix")
plt.tight_layout()
plt.show()
```
